```python
# Librerie e moduli
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
```

```python
# Lettura del file input
file_path = 'out_ann.csv'
df = pd.read_csv(file_path, delimiter='\t', usecols=["Path", "Ethnicity", "Gender"])
```

```python
# Conteggio del numero di persone per etnia 
conteggio_etnia = df["Ethnicity"].value_counts().reset_index()
conteggio_etnia.columns = ["Ethnicity", "Count"] 
conteggio_etnia = conteggio_etnia.sort_values(by="Count", ascending=False)

# Calcolo del totale delle persone e aggiunta della riga TOT
totale_persone = conteggio_etnia["Count"].sum()
totale_row = pd.DataFrame({"Ethnicity": ["TOT"], "Count": [totale_persone]})
conteggio_etnia = pd.concat([conteggio_etnia, totale_row], ignore_index=True)

# Stampa del conteggio con TOT
print("\nNumero di persone per etnia:")
display(conteggio_etnia.style.hide(axis='index'))

# Grafico con TOT
print("\nIstogramma (con TOT):")
conteggio_etnia.set_index("Ethnicity").plot(kind='bar', color='#D3D3F7', title='Numero di persone per Etnia')
plt.ylabel('Count')
plt.xlabel('Ethnicity')
plt.xticks(rotation=45)
plt.show()

# Versione della tabella senza la riga TOT
conteggio_etnia_senza_tot = conteggio_etnia[conteggio_etnia["Ethnicity"] != "TOT"]

# Grafico senza TOT
print("\nIstogramma (senza TOT):")
conteggio_etnia_senza_tot.set_index("Ethnicity").plot(kind='bar', color='#D3D3F7', title='Numero di persone per Etnia')
plt.ylabel('Count')
plt.xlabel('Ethnicity')
plt.xticks(rotation=45)
plt.show()
```

```python
# Conteggio del numero di persone per etnia e genere
conteggio_etnia_genere = df.groupby(["Ethnicity", "Gender"]).size().unstack(fill_value=0)

# Calcolo del totale per genere e aggiunta della riga TOT
totale_per_genere = conteggio_etnia_genere.sum()
totale_per_genere.name = 'TOT'
totale_row = pd.DataFrame(totale_per_genere).T
totale_row.index = ['TOT']
conteggio_etnia_genere = pd.concat([conteggio_etnia_genere, totale_row])

# Stampa del conteggio con TOT incluso
print("\nNumero di persone per etnia e genere:")
display(conteggio_etnia_genere)

# Grafico con TOT
print("\nIstogramma con TOT:")
conteggio_etnia_genere.plot(kind='bar', stacked=False, color=['pink', 'skyblue'], title='Numero di persone per Etnia e Genere')
plt.ylabel('Count')
plt.xlabel('Ethnicity')
plt.xticks(rotation=45)
plt.show()

# Versione della tabella senza la riga TOT
conteggio_etnia_genere_senza_tot = conteggio_etnia_genere.drop(index="TOT")

# Grafico senza TOT
print("\nIstogramma senza TOT:")
conteggio_etnia_genere_senza_tot.plot(kind='bar', stacked=False, color=['pink', 'skyblue'], title='Numero di persone per Etnia e Genere')
plt.ylabel('Count')
plt.xlabel('Ethnicity')
plt.xticks(rotation=45)
plt.show()
```

```python
# STATISTICHE SU MACROAREE DI MESTIERI

categorie = [
    "Management, professional, and related occupations",
    "service occupations",
    "Sales and office occupations",
    "Natural resources, construction, and maintenance occupations",
    "Production, transportation, and material moving occupations"
]

risultati = {}

for categoria in categorie:
    
    pattern = f"\\\\{categoria}"
    df_filtrato = df[df["Path"].str.contains(pattern, case=False, na=False, regex=True)]
    
    # Conteggio del numero di persone per etnia
    conteggio_etnia = df_filtrato["Ethnicity"].value_counts().reset_index()
    conteggio_etnia.columns = ["Ethnicity", "Count"]
    conteggio_etnia = conteggio_etnia.sort_values(by="Count", ascending=False)
    totale_persone = conteggio_etnia["Count"].sum()
    totale_row = pd.DataFrame({"Ethnicity": ["TOT"], "Count": [totale_persone]})
    conteggio_etnia = pd.concat([conteggio_etnia, totale_row])

    # Conteggio del numero di persone per etnia e genere
    conteggio_etnia_genere = df_filtrato.groupby(["Ethnicity", "Gender"]).size().unstack(fill_value=0)
    totale_per_genere = conteggio_etnia_genere.sum()
    totale_per_genere.name = 'TOT'
    totale_row_genere = pd.DataFrame(totale_per_genere).T
    totale_row_genere.index = ['TOT']
    conteggio_etnia_genere = pd.concat([conteggio_etnia_genere, totale_row_genere])

    risultati[categoria] = {
        "conteggio_etnia": conteggio_etnia,
        "conteggio_etnia_genere": conteggio_etnia_genere
    }

    print("----------------------------------------------------------")
    print(f"\nNumero di persone per etnia in '{categoria}':")
    display(conteggio_etnia.style.hide(axis='index'))
    
    print(f"\nNumero di persone per etnia e genere in '{categoria}':")
    display(conteggio_etnia_genere)

    # Istogramma con TOT per la categoria corrente
    print(f"\nIstogramma con TOT per le '{categoria}':")
    conteggio_etnia_genere.plot(kind='bar', stacked=False, color=['pink', 'skyblue'], title=f'Numero di persone per Etnia e Genere ({categoria})')
    plt.ylabel('N° people')
    plt.xlabel('Ethnicity')
    plt.xticks(rotation=45)
    plt.show()

    # Creazione di una versione senza la riga TOT
    conteggio_etnia_genere_senza_tot = conteggio_etnia_genere.drop(index="TOT")

    # Istogramma senza TOT per la categoria corrente
    print(f"\nIstogramma senza TOT per le '{categoria}':")
    conteggio_etnia_genere_senza_tot.plot(kind='bar', stacked=False, color=['pink', 'skyblue'], title=f'Numero di persone per Etnia e Genere ({categoria})')
    plt.ylabel('N° people')
    plt.xlabel('Ethnicity')
    plt.xticks(rotation=45)
    plt.show()
```
